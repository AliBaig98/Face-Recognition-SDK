```python
# ==========================================================
# Imports
# ==========================================================

from pathlib import Path
import shutil

import cv2
from tqdm import tqdm
```

```python
# ==========================================================
# Paths
# ==========================================================

PROJECT_ROOT = Path.cwd().parent

RAW_DIR = PROJECT_ROOT / "data" / "raw" / "lfw"

PROCESSED_DIR = PROJECT_ROOT / "data" / "processed" / "lfw_filtered"

PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

print(RAW_DIR)
print(PROCESSED_DIR)
```

```python
# ==========================================================
# Find Corrupted Images
# ==========================================================

corrupted = []

for person in tqdm(list(RAW_DIR.iterdir())):

    if not person.is_dir():
        continue

    for image in person.glob("*.jpg"):

        img = cv2.imread(str(image))

        if img is None:
            corrupted.append(image)

print(f"Corrupted Images : {len(corrupted)}")
```

```python
# ==========================================================
# Keep Identities Having >=5 Images
# ==========================================================

MIN_IMAGES = 5

selected = []

for person in RAW_DIR.iterdir():

    if not person.is_dir():
        continue

    images = list(person.glob("*.jpg"))

    if len(images) >= MIN_IMAGES:
        selected.append(person)

print(f"Selected Identities : {len(selected)}")
```

```python
# ==========================================================
# Create Filtered Dataset
# ==========================================================

for person in tqdm(selected):

    destination = PROCESSED_DIR / person.name

    destination.mkdir(exist_ok=True)

    for image in person.glob("*.jpg"):

        shutil.copy(image, destination / image.name)

print("Dataset Created Successfully")
```
