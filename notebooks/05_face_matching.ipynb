```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd().parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
```

```python
from src.data.face_detector import FaceDetector
from src.models.face_embedder import FaceEmbedder
from src.inference.face_matcher import FaceMatcher
```

```python
detector = FaceDetector()
embedder = FaceEmbedder()
matcher = FaceMatcher()
```

```python
img1 = PROJECT_ROOT / "data/processed/lfw_filtered/George_W_Bush/George_W_Bush_0001.jpg"

img2 = PROJECT_ROOT / "data/processed/lfw_filtered/George_W_Bush/George_W_Bush_0002.jpg"
```

```python
face1 = detector.detect_face(img1)
face2 = detector.detect_face(img2)

emb1 = embedder.get_embedding(face1)
emb2 = embedder.get_embedding(face2)

score = matcher.compare(emb1, emb2)

print(f"Similarity Score: {score:.4f}")
```

```python
img3 = PROJECT_ROOT / "data/processed/lfw_filtered/Colin_Powell/Colin_Powell_0001.jpg"

face3 = detector.detect_face(img3)

emb3 = embedder.get_embedding(face3)

score = matcher.compare(emb1, emb3)

print(f"Similarity Score: {score:.4f}")
```
